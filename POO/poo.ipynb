```
#Classe
class Casa:
  def __init__ (self,cor, quartos, banheiros):
    self.cor = cor
    self.quartos = quartos
    self.banheiros = banheiros

  def mostrar_cor(self):
    print(f'Acor da casa e {self.cor}')

  def mostrar_quartos(self):
    print(f'A casa tem {self.quartos} quartos')

  def mostrar_banheiros(self):
    print(f'A casa tem {self.banheiros} banheiros')

  def adicionar_quarto(self):
    self.quartos +=1
    print(f'Está casa tem {self.quartos} quartos')

  def pintar_casa(self, nova_cor):
    self.cor = nova_cor
    print(f'A casa agora é {self.cor}')

casa1 = Casa('azul', 4, 2)
casa2 = Casa('amarela', 6, 1)

print('\nCasa 1:')
casa1.mostrar_cor()
casa1.mostrar_quartos()
casa1.mostrar_banheiros()
casa1.adicionar_quarto()
casa1.pintar_casa('vermelha')

print('\nCasa 2:')
casa2.mostrar_cor()
casa2.mostrar_quartos()
casa2.mostrar_banheiros()
```

```
#Classe Carro
class Carro:
  def __init__(self, cor, ano):
     self.cor = cor
     self.ano = ano
     self.ligado = True
     self.seta = None

  def informacoes(self):
    print(f'A cor do carro é {self.cor}')
    print(f'O ano do carro {self.ano}')

  def ligar(self):
    if not self.ligado:
      self.ligado = True
      print(f'O carro foi ligado')
    else:
      print(f'O carro ja estava ligado!')

  def desligar(self):
    if self.ligado:
      self.ligado = False
      print(f'O carro foi desligado!')
    else:
      print(f'O carro ja estava desligado!')

  def ligar_seta(self, direcao):
    if not self.ligado:
      print(f'Ligue o carro antes de dar seta!')
    else:
      self.seta = direcao
      print(f'Seta ligada para a {self.seta}')

carro1 = Carro('Branco', 2012)
carro1.informacoes()
carro1.desligar()
carro1.ligar_seta('esquerda')
```

```
#Classe Empresa
class Pessoa:
  def __init__ (self, nome, idade, cargo):
    self.nome = nome
    self.idade = idade
    self.cargo = cargo

  def informacoes(self):
    print(f'Nome: {self.nome}')
    print(f'Idade: {self.idade}')
    print(f'Cargo: {self.cargo}')

  def promover(self, novo_cargo):
    print(f'{self.nome} foi promovido (a) para o cargo {novo_cargo}')
    self.cargo = novo_cargo

  def atualizar_idade(self, nova_idade):
    if nova_idade > self.idade:
      print(f'Idade deixa de ser {self.idade} anos e foi atualizada para {nova_idade} anos')
      self.idade = nova_idade
    else:
      print(f'Não foi possivel atualizar! A nova idade tem que ser maior que a anterior!')

colaborador1 = Pessoa('Julio', 21,'Auxiliar')
colaborador2 = Pessoa('Ana', 24,'Junior')

colaborador2.atualizar_idade(30)

colaborador1.promover('senior')

colaborador1.informacoes()

```

```
#Herança Multipla

#clss Avô
class Animal():
  def __init__(self, nome):
    self.nome = nome

  def falar(self):
    print(f'{self.nome} está falando!')


#Classe pai
class Predator(Animal):
  def cacando(self):
    print(f'O Animal {self.nome} está caçando!')

class Presa(Animal):
  def fugindo(self):
    print(f'O Animal {self.nome} está fugindo!')

#Classe Filho
class Coelho(Presa):
  pass

class Tigre(Predator):
  pass

class Golfinho(Predator, Presa):
  pass

coelho1 = Coelho('Bunny')
tigre1 = Tigre('Leo')
golfinho1 = Golfinho('Billy')

coelho1.fugindo()
tigre1.cacando()

golfinho1.cacando()
golfinho1.fugindo()
```

```
#Sistema de Escola

class Escola():
  def __init__(self, nome, idade, status):
    self.nome = nome
    self.idade = idade
    self.status = status

  def Apresentar(self):
      print(f'Meu nome é {self.nome}')

  def verificar_status(self):
    print(f'Status: {"ATIVO" if self.status else "INATIVO"}')

class Aluno(Escola):
  def __init__(self, nome, idade, status, ano):
    super().__init__(nome, idade, status)
    self.ano = ano

  def Apresentar(self):
    super().Apresentar()
    print(f'Eu sou aluno(a) da escola')

class Professor(Escola):
  def __init__(self, nome, idade, status, materia):
    super().__init__(nome, idade, status)
    self.materia = materia

  def Apresentar(self):
    super().Apresentar()
    print(f'Eu sou professor(a) da escola')

class Assistente(Escola):
  def __init__(self, nome, idade, status, bloco):
    super().__init__(nome, idade, status)
    self.bloco = bloco

  def Apresentar(self):
    super().Apresentar()
    print(f'Eu sou assistente(a) da escola')


a1 = Aluno(nome='Marcos', idade=21, status=True, ano=8 )
p1 = Professor(nome='Roberto', idade=34, status=True, materia='Geometria')
as1 = Assistente(nome='Maria', idade=28, status=False, bloco='A')

a1.Apresentar()
a1.verificar_status()

p1.Apresentar()
p1.verificar_status()

as1.Apresentar()
as1.verificar_status()

```

```
#Polymorphism
class Personagens():
  def falar(self):
    print('Eu sou um personagem!')

class Guerreiro(Personagens):
    def falar(self):
      print('Eu sou um guerreiro forte e destemido!')

class Mago(Personagens):
  def falar(self):
      print('Eu sou um mago sábio e poderoso!')

class Arqueiro(Personagens):
   def falar(self):
      print('Eu sou um arqueiro rapido e destemido!')

#Criar objetos:

personagens = [Guerreiro(), Mago(), Arqueiro()]

for p in personagens:
  p.falar()
```

```
#Polymorphism Animais

class Cachorro:
  def emitir_som(self):
    print('Au Au!!')

class Gato:
  def emitir_som(self):
    print('Miau!!')

class Vaca:
  def emitir_som(self):
    print('Muuuu!!')

animais = [Cachorro(), Gato(), Vaca()]

for animal in animais:
  animal.emitir_som()
```

```
#Polymorphism Pessoa

class Pessoa1:
  def falar_nome(self):
    print('Meu nome é Júlio')

class Pessoa2:
  def falar_nome(self):
    print('Meu nome é Andre')

class Pessoa3:
  def falar_nome(self):
    print('Meu nome é Rogerio')

sala = [Pessoa1(), Pessoa2(), Pessoa3()]

for pessoas in sala:
  pessoas.falar_nome()
```

```
# Criando uma fábrica de Motores com Aggregation

class Motor:
  def __init__(self, marca, potencia):
    self.marca = marca
    self.potencia = potencia

class Carro:
  def __init__(self):
    self.motores = []

  def adicionar_motor(self, motor):
    self.motores.append(motor)

  def listar_motores(self):
    for motor in self.motores:
      print(f'Marca: {motor.marca}, Potencia: {motor.potencia}')

#Criar motores

motor_v6 = Motor('Ford', 300)
motor_v8 = Motor('Ferrari', 650)
motor_v12 = Motor('Fiat', 800)

#Criar carro e adicionar motor
carro = Carro()
carro.adicionar_motor(motor_v6)
carro.adicionar_motor(motor_v8)
carro.adicionar_motor(motor_v12)

#Listar Motores
carro.listar_motores()

```

```
class Livro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor

class Biblioteca:
    def __init__(self):
        self.livros = []

    def adicionar_livro(self, livro):
        self.livros.append(livro)

    def listar_livros(self):
        for livro in self.livros:
            print(f'Título: {livro.titulo}, Autor: {livro.autor}')

# Criando Livros
livro1 = Livro('O Senhor dos Anéis', 'J.R.R. Tolkien')
livro2 = Livro('Harry Potter', 'J.K. Rowling')
livro3 = Livro('Dom Quixote', 'Miguel de Cervantes')

# Adicionando Livros a Biblioteca
biblioteca = Biblioteca()
biblioteca.adicionar_livro(livro1)
biblioteca.adicionar_livro(livro2)
biblioteca.adicionar_livro(livro3)

# Listando livros
biblioteca.listar_livros()
```
