# POO


```
#Classe
class Casa:
  def __init__ (self,cor, quartos, banheiros):
    self.cor = cor
    self.quartos = quartos
    self.banheiros = banheiros

  def mostrar_cor(self):
    print(f'Acor da casa e {self.cor}')

  def mostrar_quartos(self):
    print(f'A casa tem {self.quartos} quartos')

  def mostrar_banheiros(self):
    print(f'A casa tem {self.banheiros} banheiros')

  def adicionar_quarto(self):
    self.quartos +=1
    print(f'Está casa tem {self.quartos} quartos')

  def pintar_casa(self, nova_cor):
    self.cor = nova_cor
    print(f'A casa agora é {self.cor}')

casa1 = Casa('azul', 4, 2)
casa2 = Casa('amarela', 6, 1)

print('\nCasa 1:')
casa1.mostrar_cor()
casa1.mostrar_quartos()
casa1.mostrar_banheiros()
casa1.adicionar_quarto()
casa1.pintar_casa('vermelha')

print('\nCasa 2:')
casa2.mostrar_cor()
casa2.mostrar_quartos()
casa2.mostrar_banheiros()
```
