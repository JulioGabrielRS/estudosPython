```
#aritmética
numero1 = 10
numero2 = 2

print(numero1 + numero2)#soma
print(numero1 - numero2)#subtracao
print(numero1 * numero2)#multiplicacao
print(numero1 / numero2)#divisao
print(numero1 // numero2)#divisao inteira
print(numero1 % numero2)#resto da divisao
print(numero1 ** numero2)#potencia
```
